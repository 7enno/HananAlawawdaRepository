## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*


To ensure the recommendations do not go stale, the model's Mean Absolute Error (MAE) should be monitored on a rolling 30-day basis.

**Retrain Triggers:**
1.  A confirmed major Google Search Core Update is rolled out.
2.  The measured, rolling MAE degrades by more than 15% from our baseline validation score.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences

# Establishing the baseline threshold for the monitoring system
baseline_honest_mae = 0.00207
drift_threshold = baseline_honest_mae * 1.15

print(f"Baseline MAE: {baseline_honest_mae}")
print(f"🚨 Retrain Trigger: Alert data science team if rolling MAE exceeds {drift_threshold:.5f}")
```
